# Задача 1: Запись списка в файл

Инструкция

Напиши функцию write_list_to_file(filename, items), которая:
- Принимает список строк
- Записывает каждую строку в файл (каждая с новой строки)

````python
def write_list_to_file(filename: str, items: list):
    """
    Записывает список строк в файл.

    Args:
        filename (str): Имя файла
        items (list): Список строк
    """
    # TODO: Напишите код здесь
    with open(filename, "w") as f:
        for item in items:
            f.write(item + "\n")

# Тесты
write_list_to_file("test_list.txt", ["apple", "banana", "cherry"])

with open("test_list.txt", "r") as f:
    lines = f.readlines()

assert len(lines) == 3, "Тест 1 не пройден"
assert lines[0].strip() == "apple", "Тест 2 не пройден"
print("Все тесты пройдены!")
````

# Задача 2: Прочитать числа из файла

Инструкция

Напиши функцию read_numbers_from_file(filename), которая:
- Читает файл, где каждая строка — число
- Возвращает список чисел (int)

````python
def read_numbers_from_file(filename: str) -> list:
    """
    Читает числа из файла.

    Args:
        filename (str): Имя файла

    Returns:
        list: Список чисел
    """
    # TODO: Напишите код здесь
    with open(filename, "r") as f:
        lines = f.readlines()
        numbers = [int(line.strip()) for line in lines if line.strip()]
    return numbers

# Создадим тестовый файл
with open("numbers.txt", "w") as f:
    f.write("10\n20\n30\n")

# Тесты
numbers = read_numbers_from_file("numbers.txt")
assert numbers == [10, 20, 30], "Тест не пройден"
print("Все тесты пройдены!")
````

# Задача 3: Добавить строку в CSV

Инструкция

Напиши функцию add_row_to_csv(filename, row_data), которая:
- Принимает словарь (например, {"Name": "Eve", "Age": 22})
- Добавляет строку в существующий CSV

````python
import pandas as pd

def add_row_to_csv(filename: str, row_data: dict):
    """
    Добавляет строку в CSV файл.

    Args:
        filename (str): Имя файла
        row_data (dict): Данные строки
    """
  # TODO: Напишите код здесь
    df = pd.read_csv(filename)
    df = pd.concat([df, pd.DataFrame([row_data])])
    df.to_csv(filename)


# Создадим тестовый CSV
pd.DataFrame({"Name": ["Alice"], "Age": [25]}).to_csv("test_add.csv", index=False)

# Добавляем строку
add_row_to_csv("test_add.csv", {"Name": "Bob", "Age": 30})

# Тест
df = pd.read_csv("test_add.csv")
assert len(df) == 2, "Тест не пройден"
assert df.iloc[1]["Name"] == "Bob", "Строка не добавлена"
print("Все тесты пройдены!")
````

# Задача 4: Найти максимальное значение в CSV

Инструкция

Напиши функцию find_max_in_csv(filename, column_name), которая:
- Читает CSV
- Находит максимальное значение в указанном столбце
- Возвращает это значение

````python
import pandas as pd

def find_max_in_csv(filename: str, column_name: str):
    """
    Находит максимальное значение в столбце CSV.

    Args:
        filename (str): Имя файла
        column_name (str): Название столбца

    Returns:
        Максимальное значение
    """
  # TODO: Напишите код здесь
    df = pd.read_csv(filename)
    return df[column_name].max()

# Создадим тестовый CSV
pd.DataFrame({"Score": [85, 90, 78, 92]}).to_csv("test_max.csv", index=False)

# Тест
max_score = find_max_in_csv("test_max.csv", "Score")
assert max_score == 92, "Тест не пройден"
print("Все тесты пройдены!")
````

# Задача 5: Статистика по датасету

Инструкция

Напиши функцию dataset_stats(filename), которая:
- Читает CSV с числовыми данными
- Вычисляет среднее, минимум, максимум для всех числовых столбцов
- Возвращает словарь со статистикой

Пример вывода:

```
{
    "Age": {"mean": 27.5, "min": 25, "max": 30},
    "Score": {"mean": 86.25, "min": 78, "max": 92}
}

```




````python
import pandas as pd

def dataset_stats(filename: str) -> dict:
    """
    Вычисляет статистику по числовым столбцам CSV.

    Args:
        filename (str): Имя файла

    Returns:
        dict: Статистика {column: {mean, min, max}}
    """

  # TODO: Напишите код здесь
    df = pd.read_csv(filename)
    numeric_df = df.select_dtypes(include="number")

    stats = {}
    for col in numeric_df.columns:
        stats[col] = {
            "mean": numeric_df[col].mean(),
            "min": numeric_df[col].min(),
            "max": numeric_df[col].max()
        }
    return stats

# Создадим тестовый CSV
pd.DataFrame({
    "Age": [25, 30],
    "Score": [85, 90]
}).to_csv("test_stats.csv", index=False)

# Тест
stats = dataset_stats("test_stats.csv")
assert stats["Age"]["mean"] == 27.5, "Тест 1 не пройден"
assert stats["Score"]["max"] == 90, "Тест 2 не пройден"
print("Все тесты пройдены!")
````
