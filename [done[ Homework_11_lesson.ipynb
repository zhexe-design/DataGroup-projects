## Часть 5: Обработка пропусков

**Задача:**
1. Для каждой колонки с пропусками выбрать стратегию
2. Применить выбранную стратегию
3. Проверить что пропусков не осталось

**Стратегии:**
- Удалить колонку (если >50% пропусков)
- Удалить строки (если <5% пропусков)
- Заполнить медианой (для чисел)
- Заполнить модой (для категорий)

**Если пропусков нет — напишите об этом и переходите к следующей части**

```python
# Обработка пропусков
#
# Удаление колонки: df.drop(columns=['название'], inplace=True)
# Удаление строк: df.dropna(subset=['название'], inplace=True)
# Заполнение медианой: df['название'].fillna(df['название'].median(), inplace=True)
# Заполнение модой: df['название'].fillna(df['название'].mode()[0], inplace=True)

for col in df.columns:
    missing_ratio = df[col].isna().mean()

    if missing_ratio == 0:
        print(f"В колонке '{col}' пропусков нет.")
        continue

    if missing_ratio > 0.5:
        df.drop(columns=[col], inplace=True)
        print(f"Колонка '{col}' удалена (более 50% пропусков).")
        continue

    if missing_ratio < 0.05:
        df.dropna(subset=[col], inplace=True)
        print(f"Пропуски в колонке '{col}' удалены (<5%).")
        continue

    if pd.api.types.is_numeric_dtype(df[col]):
        df[col].fillna(df[col].median(), inplace=True)
        print(f"Пропуски в колонке '{col}' заполнены медианой.")
    else:
        df[col].fillna(df[col].mode()[0], inplace=True)
        print(f"Пропуски в колонке '{col}' заполнены модой.")
```

```python
# Проверка: пропусков не осталось
# df.isnull().sum().sum()

df.isnull().sum().sum()
```
