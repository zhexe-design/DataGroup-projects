## 4.7 Core implementation

Основная логика проекта. Разнесите её на функции. Минимум:

- функция, которая принимает вход пользователя и возвращает результат;
- если используете цепочку — каждый шаг отдельной функцией;
- обработка ошибок через `try / except` для сетевых вызовов и парсинга JSON.

Не пишите всё в одной длинной ячейке. Дробите на логические блоки и комментируйте, что каждая функция делает.


1. Главная управляющая функция (Main Orchestrator)
В коде это была бы функция def handle_user_request(user_input). В n8n это связка трех нод:

Функция приема данных (Input): Нода Telegram Trigger. Ловит входящий вебхук, извлекает текст сообщения и ID чата. При ошибке сети на этом этапе запрос просто не дойдет до системы, что безопасно для бэкенда.

Функция ядра рассуждений (Reasoning Engine): Нода AI Agent. Она принимает текст, подтягивает контекст из ноды Window Buffer Memory (функция сохранения состояния) и на основе системного промпта решает, какую функцию-инструмент вызвать дальше.

Функция возврата ответа (Output): Нода Telegram (Send Message). Принимает финальный текстовый результат от Агента и отправляет его пользователю.

2. Цепочка шагов (Prompt Chaining / Tool Execution как отдельные функции)
Каждый инструмент, подключенный к ноде AI Agent, работает как изолированная функция. В вашем проекте реализованы следующие "функции-ноды":

Функция def get_kitchen_inventory()
Нода: Google Sheets (Операция: Read rows).

Вход: Запрос Агента на чтение таблицы.

Логика: Обращается к файлу aura_inventory и забирает массив строк.

Выход: JSON-массив с текущими продуктами.

Функция def update_kitchen_inventory(product, quantity, unit)
Нода: Google Sheets (Операция: Append or Update).

Вход: Выделенные ИИ параметры (Название, Количество, Единица измерения).

Логика: Ищет совпадение по имени продукта. Если находит — обновляет количество, если нет — создает новую строку.

Выход: Статус успешной записи.

Функция def get_current_weather()
Нода: HTTP Request (Метод: GET).

Вход: URL-адрес API погоды и параметры (q=Astana).

Логика: Выполняет внешний сетевой REST-запрос.

Выход: Сырой JSON-ответ от метеослужбы.

3. Обработка ошибок (Аналог try / except в no-code)
Для сетевых вызовов (запросы к API Google, OpenAI или погоде) и парсинга данных в n8n используется встроенный механизм Error Handling, который настраивается в параметрах каждой конкретной ноды (вкладка Settings внутри ноды):

Перехват ошибок (On Error = Continue Workflow): Это полный аналог блока except. Если Google Таблицы недоступны или API погоды вернул ошибку 500, нода не «роняет» весь воркфлоу. Она передает по цепочке объект ошибки. Агент видит, что инструмент вернул сбой, и эмпатично отвечает пользователю на естественном языке: «Извините, сейчас я не могу связаться с базой данных, попробуйте чуть позже», вместо того чтобы бот просто завис.

Защита от некорректного JSON (Парсинг): Перевод нод Google Sheets в режим Map Manually с жестким описанием типов данных (значок ✨) работает как встроенный валидатор. Если ИИ пытается передать невалидный тип данных (например, текст вместо цифры в Quantity), нода n8n блокирует некорректную запись на уровне интерфейса, предотвращая поломку структуры таблицы.

> *Если у вас цепочка из нескольких шагов (Prompt Chaining), оформите каждый шаг отдельной функцией: `step_1_plan`, `step_2_extract`, `step_3_synthesize` и так далее. Финальная функция должна объединять их все в один pipeline.*

Валидность передаваемого JSON: Изначально в режиме Map Automatically модель передавала в инструмент Google Sheets избыточный технический JSON с метаданными Telegram (ID сообщения, юзернейм, даты в Unix-формате), что приводило к созданию лишних столбцов и замусориванию таблицы. Перевод ноды в режим Map Manually полностью решил эту проблему: теперь модель формирует и передает строго валидный, очищенный JSON-пакет параметров (Product, Quantity, Unit), соответствующий структуре базы данных.

Точность определения целевых метрик (Классификация интентов и параметров):

Извлечение сущностей (Entity Extraction): Благодаря точечным промптам в Tool Parameters (Output Format Control), модель успешно извлекает текстовые и числовые значения. Например, из фразы «добавь полтора литра кефира» в поле Quantity корректно передается число 1.5, а в Product — существительное Кефир в именительном падеже.

Где модель ошибается и почему:
Проблема множественного ввода: При отправке пользователем длинного неструктурированного перечисления (например, «добавь в холодильник молоко, упаковку яиц, творог и сыр») в рамках одного сообщения, агент на базе GPT-4o в стандартном режиме n8n пытается записать всё одной строкой или заполняет только первую позицию, теряя остальные.

Причина: Ограничение логики работы базового инструмента Google Sheets в n8n, который по умолчанию ожидает одну итерацию записи на один вызов.

Решение в домашке: Проблема была успешно купирована добавлением жесткого ограничения в системное описание самого инструмента (Tool Description): «If the user lists multiple items, you MUST call this tool separately for EACH item, one by one». Это заставило Агента запускать цикл последовательных вызовов для каждой позиции.

Что можно улучшить:
Добавление Few-Shot примеров для сложных сценариев: Для повышения отказоустойчивости ИИ при обработке смешанного языка или специфического сленга, планируется внедрить текстовый блок с 3–5 примерами идеального извлечения параметров (input/output) прямо в системное сообщение Агента.

Слой жесткой валидации данных (Data Validation Node): Сейчас система полностью доверяет "разумности" LLM. Чтобы исключить человеческий фактор и галлюцинации ИИ (например, если модель случайно запишет текст в числовое поле Quantity), перед нодой Google Sheets необходимо добавить ноду Code (JavaScript). Она будет выполнять программную проверку типов данных (typeof quantity === 'number') и возвращать ошибку в случае сбоя, действуя как классический блок try/except.

Оптимизация Prompt Chaining (Цепочек вызовов): При выполнении комбинированных запросов (например, запись расхода + создание события в календаре) воркфлоу тратит около 3–5 секунд на последовательное выполнение. В будущем логику можно оптимизировать через параллельный запуск независимых нод для сокращения задержки (Latency).
