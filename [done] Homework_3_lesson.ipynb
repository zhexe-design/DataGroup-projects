# Задача 1: Проверка пароля (функции + условия)

Инструкция

Напиши функцию is_valid_password(password), которая:
1. Проверяет, что длина пароля >= 6 символов
2. Возвращает True если валиден, иначе False

```python
def is_valid_password(password: str) -> bool:
    """
    Проверяет валидность пароля.

    Args:
        password (str): Пароль

    Returns:
        bool: True если валиден, False иначе
    """
    # TODO: Напиши код здесь
    if len(password) >= 6:
        return True
    else:
        return False

# Тесты
assert is_valid_password("123456") == True, "Тест 1 не пройден"
assert is_valid_password("12345") == False, "Тест 2 не пройден"
assert is_valid_password("") == False, "Тест 3 не пройден"
print("Все тесты пройдены!")
```

# Задача 2: Безопасное преобразование в число

Инструкция

Напиши функцию safe_int_convert(value), которая:
1. Преобразует value в int
2. Если не получается возвращает 0

```python
def safe_int_convert(value):
    """
    Безопасно преобразует значение в int.

    Args:
        value: Значение для преобразования

    Returns:
        int: Число или 0
    """
    # TODO: Напиши код здесь
    try:
        return int(value)
    except:
        return 0


# Тесты
assert safe_int_convert("123") == 123, "Тест 1 не пройден"
assert safe_int_convert("abc") == 0, "Тест 2 не пройден"
assert safe_int_convert(None) == 0, "Тест 3 не пройден"
print("Все тесты пройдены!")
```

# Задача 3: Получить первый элемент списка

Инструкция

Напиши функцию get_first_element(lst), которая:
1. Возвращает первый элемент списка
2. Если список пустой возвращает None

```python
def get_first_element(lst):
    """
    Возвращает первый элемент списка.

    Args:
        lst (list): Список

    Returns:
        Первый элемент или None
    """
    # TODO: Напиши код здесь
    if lst:
        return lst[0]
    else:
      return None


# Тесты
assert get_first_element([1, 2, 3]) == 1, "Тест 1 не пройден"
assert get_first_element([]) is None, "Тест 2 не пройден"
assert get_first_element(["hello", "world"]) == "hello", "Тест 3 не пройден"
print("Все тесты пройдены!")
```

# Задача 4: Подсчёт чётных чисел

Инструкция

Напиши функцию count_even_numbers(numbers), которая:
1. Считает количество чётных чисел в списке
2. Если numbers не список возвращает 0

```python
def count_even_numbers(numbers):
    """
    Считает количество чётных чисел.

    Args:
        numbers: Список чисел

    Returns:
        int: Количество чётных чисел
    """
    # TODO: Напиши код здесь
    count = 0

    try:
        for num in numbers:
            num = int(num)
            if num % 2 == 0:
                count += 1
    except:
        return 0

    return count


# Тесты
assert count_even_numbers([1, 2, 3, 4, 5, 6]) == 3, "Тест 1 не пройден"
assert count_even_numbers([1, 3, 5]) == 0, "Тест 2 не пройден"
assert count_even_numbers("не список") == 0, "Тест 3 не пройден"
print("Все тесты пройдены!")
```
