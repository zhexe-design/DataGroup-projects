## Задание 1: Supervised Learning — Классификация (+3 балла)

Обучим простую модель на датасете **Iris** (цветы ириса).

```python
# Импорт библиотек
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
import pandas as pd

# Загрузка датасета
iris = load_iris()
X = iris.data  # Признаки: длина/ширина лепестков и чашелистиков
y = iris.target  # Целевая переменная: вид ириса (0, 1, 2)

# Посмотрим на данные
print("Размер X:", X.shape)
print("Размер y:", y.shape)
print("\nПервые 5 строк X:")
print(X[:5])
print("\nПервые 5 значений y:")
print(y[:5])
print("\nКлассы (виды ирисов):", iris.target_names)
```

```python
# Разделение на train и test
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

print("Train set:", X_train.shape)
print("Test set:", X_test.shape)
```

```python
# Обучение модели KNN (K-Nearest Neighbors)
model = KNeighborsClassifier(n_neighbors=3)
model.fit(X_train, y_train)

# Предсказание
y_pred = model.predict(X_test)

# Оценка точности
accuracy = accuracy_score(y_test, y_pred)
print(f"Точность модели: {accuracy:.2f}")
```

## Задание 2: Unsupervised Learning — Кластеризация (+3 балла)

Используем тот же датасет Iris, но **без меток** (без y). Попробуем сами найти группы (кластеры).

```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

# Используем только 2 признака для визуализации
X_2d = iris.data[:, :2]  # Берем первые 2 признака (sepal length, sepal width)

print("Размер X_2d:", X_2d.shape)
print("Первые 5 строк:")
print(X_2d[:5])
```

```python
# Кластеризация с KMeans (3 кластера)
kmeans = KMeans(n_clusters=3, random_state=42)
clusters = kmeans.fit_predict(X_2d)

print("Найденные кластеры:")
print(clusters[:20])
```

```python
# Визуализация кластеров
plt.figure(figsize=(10, 6))
plt.scatter(X_2d[:, 0], X_2d[:, 1], c=clusters, cmap='viridis', s=50, alpha=0.7)
plt.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1],
            c='red', marker='X', s=200, label='Centroids')
plt.xlabel('Sepal Length')
plt.ylabel('Sepal Width')
plt.title('K-Means Clustering (Unsupervised)')
plt.legend()
plt.show()
```

## Задание 3: Сравнение Supervised vs Unsupervised (+2 балла)

Сравните результаты кластеризации с реальными метками.

```python
# Визуализация: Кластеры vs Реальные классы
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# Слева: кластеры KMeans (Unsupervised)
axes[0].scatter(X_2d[:, 0], X_2d[:, 1], c=clusters, cmap='viridis', s=50, alpha=0.7)
axes[0].set_title('Unsupervised: KMeans Clusters')
axes[0].set_xlabel('Sepal Length')
axes[0].set_ylabel('Sepal Width')

# Справа: реальные классы (Supervised)
axes[1].scatter(X_2d[:, 0], X_2d[:, 1], c=y, cmap='viridis', s=50, alpha=0.7)
axes[1].set_title('Supervised: Real Classes')
axes[1].set_xlabel('Sepal Length')
axes[1].set_ylabel('Sepal Width')

plt.tight_layout()
plt.show()
```
