# Задача 1: Получить email пользователя

Инструкция

Напиши функцию get_user_email(user_id), которая:
- Загружает пользователя по ID
- Возвращает его email
- Если ошибка возвращает None

```python
import requests

def get_user_email(user_id: int) -> str:
    """
    Получает email пользователя.

    Args:
        user_id (int): ID пользователя

    Returns:
        str: Email или None
    """
    # TODO: Напиши код здесь
    try:
        response = requests.get(f"https://jsonplaceholder.typicode.com/users/{user_id}")
        if response.status_code != 200:
            return None

        user = response.json()
        return user.get("email")
    except:
        return None


# Тесты
assert get_user_email(1) == "Sincere@april.biz", "Тест 1 не пройден"
assert get_user_email(5) == "Lucio_Hettinger@annie.ca", "Тест 2 не пройден"
assert get_user_email(999) is None, "Тест 3 не пройден"
print("Все тесты пройдены!")
```

# Задача 2: Получить заголовок поста

Инструкция

Напиши функцию get_post_title(post_id), которая:
- Загружает пост по ID
- Возвращает его заголовок
- Если статус не 200 — возвращает "Post not found"

```python
import requests

def get_post_title(post_id: int) -> str:
    """
    Получает заголовок поста.

    Args:
        post_id (int): ID поста

    Returns:
        str: Заголовок или сообщение об ошибке
    """
    # TODO: Напиши код здесь
    try:
        response = requests.get(f"https://jsonplaceholder.typicode.com/posts/{post_id}")

        if response.status_code != 200:
            return "Post not found"

        post = response.json()
        return post.get("title")

    except:
        return "Post not found"


# Тесты
title1 = get_post_title(1)
assert "sunt aut" in title1.lower(), "Тест 1 не пройден"

title_invalid = get_post_title(99999)
assert title_invalid == "Post not found", "Тест 2 не пройден"
print("Все тесты пройдены!")
```

# Задача 3: Создать комментарий

Инструкция

Напиши функцию create_comment(post_id, email, body), которая:
- Отправляет POST запрос для создания комментария
- Возвращает ID нового комментария
- Если ошибка — возвращает None

```python
import requests

def create_comment(post_id: int, email: str, body: str) -> int:
    """
    Создаёт комментарий к посту.

    Args:
        post_id (int): ID поста
        email (str): Email автора
        body (str): Текст комментария

    Returns:
        int: ID комментария или None
    """
    # TODO: Напиши код здесь
    try:
        response = requests.post(
            "https://jsonplaceholder.typicode.com/comments",
            json={
                "postId": post_id,
                "email": email,
                "body": body
            }
        )

        if response.status_code not in (200, 201):
            return None

        data = response.json()
        return data.get("id")

    except:
        return None



# Тесты
cid1 = create_comment(1, "test@example.com", "Great post!")
assert cid1 is not None, "Тест 1 не пройден"

cid2 = create_comment(5, "user@mail.com", "Thanks!")
assert isinstance(cid2, int), "Тест 2 не пройден"
print("Все тесты пройдены!")
```

# Задача 4: Получить среднюю длину титлов

Инструкция

Напиши функцию avg_title_length(limit), которая:
- Загружает N постов (параметр _limit=limit)
- Вычисляет среднюю длину заголовков
- Возвращает среднее (int) или 0 если ошибка

```python
import requests

def avg_title_length(limit: int) -> int:
    """
    Вычисляет среднюю длину заголовков постов.

    Args:
        limit (int): Количество постов

    Returns:
        int: Средняя длина или 0
    """
    # TODO: Напиши код здесь
    try:
        resp = requests.get(
            "https://jsonplaceholder.typicode.com/posts",
            params={"_limit": limit}
        )

        if resp.status_code != 200:
            return 0

        posts = resp.json()
        if not posts:
            return 0

        total = sum(len(p.get("title", "")) for p in posts)
        return total // len(posts)

    except:
        return 0


# Тесты
avg1 = avg_title_length(5)
assert isinstance(avg1, int), "Тест 1 не пройден"
assert avg1 > 0, "Тест 2 не пройден"

avg2 = avg_title_length(10)
assert avg2 > 0, "Тест 3 не пройден"
print("Все тесты пройдены!")
```
