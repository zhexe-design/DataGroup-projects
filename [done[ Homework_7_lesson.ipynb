# Задача 1: Очистка списка оценок (Логика + Циклы)
Инструкция:

Напиши функцию clean_scores(scores), которая:
1. Удаляет None и строки, которые нельзя превратить в число.
2. Удаляет числа меньше 0 и больше 100 (ошибочные данные).
3. Возвращает список float.

```python
def clean_scores(scores: list) -> list:
    """
    Очищает список оценок.
    Оставляет только числа от 0 до 100.
    """
    # TODO: Напиши код здесь
    cleaned = []
    for item in scores:
        if item is None:
            continue
        try:
            number = float(item)
        except (ValueError, TypeError):
            continue

        if 0 <= number <= 100:
            cleaned.append(number)

    return cleaned

# Тесты для Задачи 1
raw_scores = [95, "105", None, -5, "80.5", "text", 40]
cleaned = clean_scores(raw_scores)
assert len(cleaned) == 3, "Ошибка: должно остаться 3 значения (95, 80.5, 40)"
assert 105 not in cleaned, "Ошибка: 105 больше 100"
assert -5 not in cleaned, "Ошибка: отрицательное число"
assert 80.5 in cleaned, "Ошибка: строка '80.5' должна стать числом"
print("Задача 1 пройдена!")
```

# Задача 2: Нормализация Min-Max (Математика в коде)
Инструкция:

Напиши функцию normalize_list(data), которая реализует формулу:
- (x - min) / (max - min)

Результат округли до 2 знаков после запятой (round(x, 2))

```python
def normalize_list(data: list) -> list:
    """
    Приводит список чисел к диапазону [0, 1].
    """
    # TODO: Напиши код здесь
    min_val = min(data)
    max_val = max(data)

    # Защита от деления на ноль (когда все элементы одинаковы)
    if max_val == min_val:
        return [0.0 for _ in data]

    normalized = [
        round((x - min_val) / (max_val - min_val), 2)
        for x in data
    ]

    return normalized

# Тесты для Задачи 2
values = [10, 20, 30, 40, 50]
norm_values = normalize_list(values)
assert norm_values[0] == 0.0, "Первый элемент должен быть 0.0"
assert norm_values[-1] == 1.0, "Последний элемент должен быть 1.0"
assert norm_values[2] == 0.5, "Средний элемент должен быть 0.5"
print("Задача 2 пройдена!")
```

# Задача 3: Заполнение пропусков в Pandas

Инструкция:

Напиши функцию fill_missing_data(df), которая:
1. В колонке 'rating' заполняет пропуски СРЕДНИМ значением.
2. В колонке 'category' заполняет пропуски словом "Unknown".
3. Возвращает обновленный DataFrame.

```python
import pandas as pd

def fill_missing_data(df: pd.DataFrame) -> pd.DataFrame:
    """
    Заполняет пропуски: rating -> mean, category -> 'Unknown'
    """
    # TODO: Напиши код здесь
    # Подсказка: используй .fillna()
    mean_rating = df["rating"].mean()

    df["rating"] = df["rating"].fillna(mean_rating)
    df["category"] = df["category"].fillna("Unknown")

    return df

# Тесты для Задачи 3
df_test = pd.DataFrame({
    "rating": [5, None, 3, None],
    "category": ["A", "B", None, "A"]
})
df_result = fill_missing_data(df_test.copy())

expected_mean = 4.0 # (5+3)/2
assert df_result["rating"].isna().sum() == 0, "Остались пропуски в rating"
assert df_result["rating"][1] == expected_mean, "Неверно заполнено средним"
assert df_result["category"][2] == "Unknown", "Неверно заполнена категория"
print("Задача 3 пройдена!")
```

# Задача 4: Комплексная обработка (Финал)

Инструкция:

Есть сырой список транзакций. Нужно:
1. Убрать отрицательные суммы.
2. Нормализовать оставшиеся суммы (0-1).

Вернуть список нормализованных сумм.

```python
def process_transactions(transactions: list) -> list:
    # 1. Сначала очисти (убери <= 0)
    # 2. Потом нормализуй то, что осталось
    # TODO: Напиши код
    cleaned = [x for x in transactions if x > 0]

    # 2. Нормализация Min-Max
    min_val = min(cleaned)
    max_val = max(cleaned)

    if min_val == max_val:
        return [0.0 for _ in cleaned]

    normalized = [
        round((x - min_val) / (max_val - min_val), 2)
        for x in cleaned
    ]

    return normalized

# Тесты для Задачи 4
txs = [100, -50, 200, 0, 300]
# После очистки: [100, 200, 300]
# Min=100, Max=300
# 100 -> 0.0
# 200 -> 0.5
# 300 -> 1.0
processed = process_transactions(txs)
assert len(processed) == 3, "Ошибка фильтрации"
assert processed == [0.0, 0.5, 1.0], "Ошибка нормализации"
print("Задача 4 пройдена!")
```
