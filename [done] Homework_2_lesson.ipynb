# ЗАДАЧА 1: Очистка строки (строки)

Инструкция:

Напиши функцию clean_text(text), которая:
1. Убирает пробелы в начале и конце
2. Приводит текст к нижнему регистру
3. Заменяет символ "_" на пробел " "

```python
from typing import Optional

def clean_text(text: str) -> str:
    """
    Очищает текст: убирает пробелы, приводит к нижнему регистру,
    заменяет '_' на ' '.

    Args:
        text (str): Исходная строка

    Returns:
        str: Очищенная строка
    """
    # ----- НАЧАЛО ВАШЕГО КОДА -----
    return text.strip().lower().replace("_", " ")
    # ----- КОНЕЦ ВАШЕГО КОДА -----

# --- Тесты ---
print("--- Тесты Задачи 1 ---")
assert clean_text("  Hello_World  ") == "hello world", "Тест 1 не пройден"
assert clean_text("PYTHON_IS_GREAT") == "python is great", "Тест 2 не пройден"
assert clean_text("  Data_Science  ") == "data science", "Тест 3 не пройден"
assert clean_text("no_changes_needed") == "no changes needed", "Тест 4 не пройден"
print("Все тесты для Задачи 1 пройдены!")
```

# ЗАДАЧА 2: Фильтрация списка (списки)

Инструкция:

Напиши функцию filter_positive(numbers), которая:
1. Принимает список чисел
2. Возвращает НОВЫЙ список только с положительными числами (> 0)
3. Сортирует результат по возрастанию

Подсказка: Используй list comprehension и .sort() или sorted()


```python
from typing import List

def filter_positive(numbers: List[int]) -> List[int]:
    """
    Фильтрует и сортирует положительные числа.

    Args:
        numbers (list): Список чисел

    Returns:
        list: Отсортированный список положительных чисел
    """
    # ----- НАЧАЛО ВАШЕГО КОДА -----
    result = [num for num in numbers if num > 0]
    return sorted(result)
    # ----- КОНЕЦ ВАШЕГО КОДА -----

# --- Тесты ---
print("\n--- Тесты Задачи 2 ---")
assert filter_positive([3, -1, 5, 0, -2, 8]) == [3, 5, 8], "Тест 1 не пройден"
assert filter_positive([-10, -5, 0]) == [], "Тест 2 не пройден"
assert filter_positive([7, 2, 9, 1]) == [1, 2, 7, 9], "Тест 3 не пройден"
assert filter_positive([]) == [], "Тест 4 (пустой список) не пройден"
print("Все тесты для Задачи 2 пройдены!")
```

# ЗАДАЧА 3: Подсчёт слов (строки + словари)

Инструкция:

Напиши функцию count_words(text), которая:
1. Принимает строку
2. Приводит строку к нижнему регистру
3. Разбивает на слова (по пробелам)
4. Возвращает словарь: {"слово": количество_вхождений}

Подсказка:
1. text.lower().split()
2. Пройдись циклом по списку слов
3. Проверяй, есть ли слово в словаре. Если нет - добавь (со значением 1), если есть - увеличь счетчик (+1)


```python
from typing import Dict

def count_words(text: str) -> Dict[str, int]:
    """
    Подсчитывает количество вхождений каждого слова.

    Args:
        text (str): Исходная строка

    Returns:
        dict: Словарь {слово: количество}
    """
    # ----- НАЧАЛО ВАШЕГО КОДА -----
    words = text.lower().split()
    result = {}

    for word in words:
        if word in result:
            result[word] += 1
        else:
            result[word] = 1

    return result
    # ----- КОНЕЦ ВАШЕГО КОДА -----

# --- Тесты ---
print("\n--- Тесты Задачи 3 ---")
result1 = count_words("apple banana apple")
assert result1 == {"apple": 2, "banana": 1}, "Тест 1 не пройден"

result2 = count_words("Python python PYTHON")
assert result2 == {"python": 3}, "Тест 2 не пройден"

result3 = count_words("data science data")
assert result3 == {"data": 2, "science": 1}, "Тест 3 не пройден"

result4 = count_words("")
assert result4 == {}, "Тест 4 (пустая строка) не пройден"
print("Все тесты для Задачи 3 пройдены!")
```

# ЗАДАЧА 4: Объединение словарей (словари)

Инструкция:

Напиши функцию merge_configs(config1, config2), которая:
1. Принимает два словаря
2. Объединяет их (ключи из config2 перезаписывают config1, если совпадают)
3. Возвращает НОВЫЙ словарь (не изменяет исходные!)

Подсказка:
1. Создай .copy() первого словаря
2. Используй .update() для добавления/обновления из второго ИЛИ используй {**dict1, **dict2}


```python
from typing import Dict, Any

def merge_configs(config1: Dict[str, Any], config2: Dict[str, Any]) -> Dict[str, Any]:
    """
    Объединяет два словаря (config2 перезаписывает config1).

    Args:
        config1 (dict): Первый словарь
        config2 (dict): Второй словарь

    Returns:
        dict: Объединённый словарь
    """
    # ----- НАЧАЛО ВАШЕГО КОДА -----
    result = config1.copy()   # создаём копию, чтобы не менять оригинал
    result.update(config2)    # обновляем значениями из config2
    return result
    # ----- КОНЕЦ ВАШЕГО КОДА -----

# --- Тесты ---
print("\n--- Тесты Задачи 4 ---")
c1 = {"a": 1, "b": 2, "model": "BERT"}
c2 = {"b": 3, "c": 4, "learning_rate": 0.01}
result = merge_configs(c1, c2)
expected = {"a": 1, "b": 3, "c": 4, "model": "BERT", "learning_rate": 0.01}

assert result == expected, "Тест 1 не пройден"

# Проверка, что исходные словари не изменились
assert c1 == {"a": 1, "b": 2, "model": "BERT"}, "config1 был изменён!"
assert c2 == {"b": 3, "c": 4, "learning_rate": 0.01}, "config2 был изменён!"
print("Все тесты для Задачи 4 пройдены!")
```

# ЗАДАЧА 5 (Challenge): Топ-N элементов (словари + списки)

Инструкция:

Напиши функцию top_n_items(data, n), которая:
1. Принимает словарь data (например, {"item": количество})
2. Возвращает список топ-N *ключей* с максимальными *значениями*
3. Список должен быть отсортирован по УБЫВАНИЮ значений

Подсказка:

Используй функцию sorted() с параметрами `key` и `reverse`.

`key` должен принимать *значение* словаря (item[1]).

`sorted(data.items(), key=lambda item: item[1], reverse=True)`

Это вернет список кортежей [('key', val), ...].

Из него нужно взять срез [0:n] и извлечь только ключи (item[0]).
#

```python
from typing import Dict, Any, List

def top_n_items(data: Dict[str, Any], n: int) -> List[str]:
    """
    Возвращает топ-N ключей с максимальными значениями.

    Args:
        data (dict): Словарь {ключ: значение}
        n (int): Количество топ элементов

    Returns:
        list: Список ключей (топ-N)
    """
    # ----- НАЧАЛО ВАШЕГО КОДА -----
    sorted_items = sorted(data.items(), key=lambda item: item[1], reverse=True)
    return [key for key, _ in sorted_items[:n]]
    # ----- КОНЕЦ ВАШЕГО КОДА -----

# --- Тесты ---
print("\n--- Тесты Задачи 5 ---")
data1 = {"apple": 10, "banana": 5, "cherry": 15, "date": 1}
assert top_n_items(data1, 2) == ["cherry", "apple"], "Тест 1 не пройден"
assert top_n_items(data1, 3) == ["cherry", "apple", "banana"], "Тест 2 не пройден"

data2 = {"a": 1, "b": 3, "c": 2, "d": 5, "e": 4}
assert top_n_items(data2, 3) == ["d", "e", "b"], "Тест 3 не пройден"

# Тест, если n > len(data)
assert top_n_items(data1, 10) == ["cherry", "apple", "banana", "date"], "Тест 4 не пройден"
print("Все тесты для Задачи 5 пройдены!")
```
