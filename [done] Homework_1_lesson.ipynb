# ЗАДАЧА 1: Подсчет буквы "a"

Описание:
Напишите функцию `count_a(text)`, которая принимает строку и возвращает количество букв "a" (и "А") в ней.

Пример:

count_a("Anaconda") -> 3

```python
def count_a(text: str) -> int:
    """
    Считает количество букв 'a' и 'А' в строке text.
    """
    # ----- НАЧАЛО ВАШЕГО КОДА -----
    return text.lower().count('a')
    # ----- КОНЕЦ ВАШЕГО КОДА -----

# --- Тесты для проверки ---
# (Код ниже должен работать без ошибок, когда ваша функция готова)
print("--- Тесты для Задачи 1 ---")
print(f"'Anaconda': {count_a('Anaconda')}") # Ожидаемый: 3
print(f"'Hello, World!': {count_a('Hello, World!')}") # Ожидаемый: 0
print(f"'AbAbAbA': {count_a('AbAbAbA')}") # Ожидаемый: 4
print("-" * 20)
```

# ЗАДАЧА 2: Фильтр чисел из строки

Описание:
Напишите программу, которая:
1. Получает от пользователя строку чисел, разделенных пробелом (например: "10 5 30 15 20")
2. Получает от пользователя "пороговое" значение (N).
3. Выводит в одну строку (через пробел) только те числа из строки, которые БОЛЬШЕ, чем N.

Подсказка: Используй .split() для разделения строки на список строк, затем преобразуй каждую в int.


```python
print("\n--- Задача 2: Фильтр чисел ---")

# --- Раскомментируйте 2 строки ниже для запуска ---
input_string = input("Введите числа через пробел: ")
n_value_str = input("Введите пороговое значение N: ")

# ----- НАЧАЛО ВАШЕГО КОДА -----
try:
    n_value = int(n_value_str)
    parts = input_string.split()
    print("Здесь будет ваш результат...")
    print("Результат:", end=" ")
    for part in parts:
        if int(part) > n_value:
            print(part, end=" ")
except:
    print("Ошибка.")
print()

# ----- КОНЕЦ ВАШЕГО КОДА -----
print("-" * 20)
```


# ЗАДАЧА 3: Фильтр строк по букве "M"

Описание:
Дан список имен. Выберите из него только те, которые начинаются с буквы "M" (без учета регистра).

Пример:

['Maria', 'Alex', 'mike', 'John', 'Mila'] -> ['Maria', 'mike', 'Mila']

Подсказка: Используй .startswith() или срезы [0] и .lower()/.upper() для нечувствительности к регистру.

```python
print("\n--- Задача 3: Фильтр по букве 'M' ---")
names_list = ['Maria', 'Alex', 'mike', 'John', 'Mila', 'Mark', 'andrew', 'maggie']
print(f"Исходный список: {names_list}")

result_names = []

# ----- НАЧАЛО ВАШЕГО КОДА -----
for name in names_list:
    if name.lower().startswith('m'):
        result_names.append(name)
# ----- КОНЕЦ ВАШЕГО КОДА -----

print(f"Имена на 'M': {result_names}")
print("-" * 20)
```

# ЗАДАЧA 4: Квадраты от 1 до 10

Описание:

Вывод должен быть в одну строку через пробел.

# ЗАДАЧА 5: FizzBuzz (Классика)

Описание:
Напиши цикл, который проходит по числам от 1 до 50 (включительно).
1. Если число делится на 3 без остатка, выведи "Fizz".
2. Если число делится на 5 без остатка, выведи "Buzz".
3. Если число делится и на 3, и на 5, выведи "FizzBuzz".
4. В остальных случаях выведи само число.

Подсказка: Порядок проверок (if/elif/else) здесь КРАЙНЕ важен. Начинай с самой сложной проверки (деление на 3 и 5).

```python
print("\n--- Задача 5: FizzBuzz ---")

# ----- НАЧАЛО ВАШЕГО КОДА -----
for i in range(1, 51):
    if i % 3 == 0 and i % 5 == 0:
        print("FizzBuzz")
    elif i % 3 == 0:
        print("Fizz")
    elif i % 5 == 0:
        print("Buzz")
    else:
        print(i)
# ----- КОНЕЦ ВАШЕГО КОДА -----

print("-" * 20)
```
