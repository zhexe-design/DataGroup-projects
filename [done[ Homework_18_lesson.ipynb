## Выбор датасета

Выберите один из трех предложенных датасетов и раскомментируйте соответствующий код

```python3
from sklearn import datasets
from sklearn.model_selection import (
    train_test_split, KFold, StratifiedKFold, LeaveOneOut,
    cross_val_score, GridSearchCV
)
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import time

np.random.seed(42)
```

```python3
data = datasets.load_wine()
X, y = data.data, data.target
dataset_name = "Wine"
y = (y == 0).astype(int)

# ДАТАСЕТ 2: Digits (классы 0 и 1)
# data = datasets.load_digits()
# mask = (data.target == 0) | (data.target == 1)
# X = data.data[mask]
# y = data.target[mask]
# dataset_name = "Digits (0 vs 1)"

print(f"Датасет: {dataset_name}")
print(f"Размер: {X.shape}")
print(f"Классы: {np.unique(y)}")
```

---

## ШАГ 1-2: Анализ данных и подготовка

```python3
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)

print("Размер train:", X_train.shape)
print("Размер test:", X_test.shape)

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("Среднее по train:")
print(X_train_scaled.mean(axis=0)[:5])

print("\nStd по train:")
print(X_train_scaled.std(axis=0)[:5])

```

---

## ШАГ 3: Сравнение методов кросс-валидации

```python3
log_reg = LogisticRegression(solver="liblinear", max_iter=1000, random_state=42)

kf = KFold(n_splits=5, shuffle=True, random_state=42)
kf_scores = cross_val_score(log_reg, X_train_scaled, y_train, cv=kf, scoring="accuracy")

skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
skf_scores = cross_val_score(log_reg, X_train_scaled, y_train, cv=skf, scoring="accuracy")

X_subset = X_train_scaled[:100]
y_subset = y_train[:100]
loo = LeaveOneOut()
loo_scores = cross_val_score(log_reg, X_subset, y_subset, cv=loo, scoring="accuracy")

param_grid = {
    "C": [0.1, 1, 10, 100]
}

inner_cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
outer_cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

grid_search = GridSearchCV(
    estimator=log_reg,
    param_grid=param_grid,
    cv=inner_cv,
    scoring="accuracy"
)

nested_scores = cross_val_score(
    grid_search,
    X_train_scaled,
    y_train,
    cv=outer_cv,
    scoring="accuracy"
)

print("K-Fold CV scores:", kf_scores)
print("Средняя accuracy:", kf_scores.mean())

print("Stratified K-Fold CV scores:", skf_scores)
print("Средняя accuracy:", skf_scores.mean())

print("LOO accuracy (subset of 100):", loo_scores.mean())

print("Nested CV scores:", nested_scores)
print("Средняя accuracy (Nested CV):", nested_scores.mean())

```

## ШАГ 5: Исследование влияния параметров SVM

```python3
svm_linear = SVC(
    kernel="linear",
    C=1.0,
    random_state=42
)

svm_linear.fit(X_train_scaled, y_train)

y_train_pred_svm = svm_linear.predict(X_train_scaled)
y_test_pred_svm = svm_linear.predict(X_test_scaled)

svm_train_acc = accuracy_score(y_train, y_train_pred_svm)
svm_test_acc = accuracy_score(y_test, y_test_pred_svm)

n_support_vectors = svm_linear.n_support_
total_support_vectors = svm_linear.support_vectors_.shape[0]

log_reg = LogisticRegression(
    C=1.0,
    solver="liblinear",
    max_iter=1000,
    random_state=42
)

log_reg.fit(X_train_scaled, y_train)

y_train_pred_lr = log_reg.predict(X_train_scaled)
y_test_pred_lr = log_reg.predict(X_test_scaled)

lr_train_acc = accuracy_score(y_train, y_train_pred_lr)
lr_test_acc = accuracy_score(y_test, y_test_pred_lr)

print(f"LogReg Train accuracy: {lr_train_acc:.4f}")
print(f"LogReg Test accuracy:  {lr_test_acc:.4f}")

print("\nСравнение моделей:")
print(f"SVM        → Train: {svm_train_acc:.4f}, Test: {svm_test_acc:.4f}")
print(f"LogReg     → Train: {lr_train_acc:.4f}, Test: {lr_test_acc:.4f}")
```

---

## ШАГ 6: GridSearchCV для SVM

```python3
param_grid = [
    {
        "kernel": ["linear"],
        "C": [0.1, 1, 10, 100]
    },
    {
        "kernel": ["rbf"],
        "C": [0.1, 1, 10, 100],
        "gamma": [0.001, 0.01, 0.1, 1]
    }
]

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

svm = SVC()

grid_search = GridSearchCV(
    estimator=svm,
    param_grid=param_grid,
    cv=cv,
    scoring="accuracy",
    n_jobs=-1,
    return_train_score=True
)

grid_search.fit(X_train_scaled, y_train)

results = pd.DataFrame(grid_search.cv_results_)

top10 = results.sort_values(
    by="mean_test_score",
    ascending=False
).head(10)

display(
    top10[
        [
            "mean_test_score",
            "std_test_score",
            "mean_train_score",
            "param_kernel",
            "param_C",
            "param_gamma"
        ]
    ]
)

print("Лучшие параметры:", grid_search.best_params_)
print("Лучший CV accuracy:", grid_search.best_score_)

best_svm = grid_search.best_estimator_

best_svm.fit(X_train_scaled, y_train)

y_train_pred = best_svm.predict(X_train_scaled)
y_test_pred = best_svm.predict(X_test_scaled)

train_acc = accuracy_score(y_train, y_train_pred)
test_acc = accuracy_score(y_test, y_test_pred)

print(f"Train accuracy: {train_acc:.4f}")
print(f"Test accuracy:  {test_acc:.4f}")

cv_score = grid_search.best_score_

print("\nАнализ обобщающей способности:")
print(f"Best CV score: {cv_score:.4f}")
print(f"Test score:    {test_acc:.4f}")
print(f"Разница:       {cv_score - test_acc:.4f}")

```

## ШАГ 7: Nested Cross-Validation

```python3
outer_cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
inner_cv = StratifiedKFold(n_splits=3, shuffle=True, random_state=42)

log_reg = LogisticRegression(
    solver="liblinear",
    max_iter=1000,
    random_state=42
)

param_grid_lr = {
    "C": [0.1, 1, 10, 100]
}

grid_lr = GridSearchCV(
    estimator=log_reg,
    param_grid=param_grid_lr,
    cv=inner_cv,
    scoring="accuracy",
    n_jobs=-1
)

nested_scores_lr = cross_val_score(
    grid_lr,
    X_train_scaled,
    y_train,
    cv=outer_cv,
    scoring="accuracy"
)

print("Logistic Regression — Nested CV")
print(f"Mean accuracy: {nested_scores_lr.mean():.4f}")
print(f"Std accuracy:  {nested_scores_lr.std():.4f}")

param_grid_svm = {
    "C": [0.1, 1, 10],
    "gamma": [0.01, 0.1],
    "kernel": ["rbf"]
}


svm = SVC()

grid_svm = GridSearchCV(
    estimator=svm,
    param_grid=param_grid_svm,
    cv=inner_cv,
    scoring="accuracy",
    n_jobs=-1
)

nested_scores_svm = cross_val_score(
    grid_svm,
    X_train_scaled,
    y_train,
    cv=outer_cv,
    scoring="accuracy"
)

print("\nSVM (RBF) — Nested CV")
print(f"Mean accuracy: {nested_scores_svm.mean():.4f}")
print(f"Std accuracy:  {nested_scores_svm.std():.4f}")

print("\nСравнение Nested CV:")
print(
    f"Logistic Regression → "
    f"{nested_scores_lr.mean():.4f} ± {nested_scores_lr.std():.4f}"
)

print(
    f"SVM (RBF)           → "
    f"{nested_scores_svm.mean():.4f} ± {nested_scores_svm.std():.4f}"
)
```

## ШАГ 8: Итоговое сравнение

```python3
cv_final = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

grid_lr_final = GridSearchCV(
    estimator=LogisticRegression(
        solver="liblinear",
        max_iter=1000,
        random_state=42
    ),
    param_grid={"C": [0.1, 1, 10, 100]},
    cv=cv_final,
    scoring="accuracy",
    n_jobs=-1
)

start = time.time()
grid_lr_final.fit(X_train_scaled, y_train)
lr_train_time = time.time() - start

log_reg_best = grid_lr_final.best_estimator_

lr_test_acc = accuracy_score(
    y_test,
    log_reg_best.predict(X_test_scaled)
)

param_grid_svm_final = [
    {
        "kernel": ["linear"],
        "C": [0.1, 1, 10, 100]
    },
    {
        "kernel": ["rbf"],
        "C": [0.1, 1, 10],
        "gamma": [0.01, 0.1]
    }
]

grid_svm_final = GridSearchCV(
    estimator=SVC(),
    param_grid=param_grid_svm_final,
    cv=cv_final,
    scoring="accuracy",
    n_jobs=-1
)

start = time.time()
grid_svm_final.fit(X_train_scaled, y_train)
svm_train_time = time.time() - start

svm_best = grid_svm_final.best_estimator_

svm_test_acc = accuracy_score(
    y_test,
    svm_best.predict(X_test_scaled)
)

svm_support_vectors = svm_best.support_vectors_.shape[0]

lr_nested_mean = nested_scores_lr.mean()
lr_nested_std = nested_scores_lr.std()

svm_nested_mean = nested_scores_svm.mean()
svm_nested_std = nested_scores_svm.std()

comparison_table = pd.DataFrame({
    "Метрика": [
        "Test Accuracy",
        "Nested CV Mean",
        "Nested CV Std",
        "Время обучения (сек)",
        "Интерпретируемость",
        "Использованные опорные векторы"
    ],
    "Logistic Regression": [
        f"{lr_test_acc:.4f}",
        f"{lr_nested_mean:.4f}",
        f"{lr_nested_std:.4f}",
        f"{lr_train_time:.4f}",
        "Высокая",
        "—"
    ],
    "SVM": [
        f"{svm_test_acc:.4f}",
        f"{svm_nested_mean:.4f}",
        f"{svm_nested_std:.4f}",
        f"{svm_train_time:.4f}",
        "Низкая",
        f"{svm_support_vectors}"
    ],
    "Победитель": [
        "SVM" if svm_test_acc > lr_test_acc else "LR",
        "SVM" if svm_nested_mean > lr_nested_mean else "LR",
        "LR" if lr_nested_std < svm_nested_std else "SVM",
        "LR" if lr_train_time < svm_train_time else "SVM",
        "LR",
        "SVM"
    ]
})

comparison_table
```

### 2. Сравнение методов кросс-валидации

K-Fold: простая оценка, смещает результаты при дисбалансе.

Stratified K-Fold: сохраняет пропорции классов.

Leave-One-Out: почти несмещённая оценка.

Nested CV: честная оценка, предотвращает утечку информации.

### 3. Влияние параметров SVM

C: контролирует регуляризацию; большие значения меньше регуляризации, риск переобучения.

gamma: определяет ширину ядра; слишком большое → переобучение, слишком маленькое → недообучение.

Правильная комбинация C и gamma критична для максимальной точности SVM.

Nested CV даёт честную оценку качества моделей, показывает устойчивость LR и высокую точность SVM.

### 6. Сравнение LR и SVM

Logistic Regression: быстрое обучение, высокая интерпретируемость, стабильность.	Немного ниже точность, плохо улавливает нелинейности.
SVM (RBF): ысокая точность, учитывает нелинейные зависимости, использует опорные векторы. Медленнее обучение, низкая интерпретируемость, чувствителен к параметра
